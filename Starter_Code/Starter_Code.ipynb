## Preprocessing

```python
# Import our dependencies
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import pandas as pd
import tensorflow as tf
import numpy as np
# Import pandas and read the charity_data.csv from the provided cloud URL.
import pandas as pd
application_df = pd.read_csv("https://static.bc-edx.com/data/dl-1-2/m21/lms/starter/charity_data.csv")
application_df.head()
```

```python
# Drop the non-beneficial ID columns, 'EIN' and 'NAME'.

application_df = application_df.drop(columns=['EIN', 'NAME'])

# Display the first few rows to confirm the changes
application_df.head()
```

```python
# Determine the number of unique values in each column.

application_df.nunique()
```

```python
# Look at APPLICATION_TYPE value counts to identify and replace with "Other"

application_df['APPLICATION_TYPE'].value_counts()
```

```python
# Choose a cutoff value (e.g., fewer than 500 occurrences)
application_type_counts = application_df['APPLICATION_TYPE'].value_counts()

# Create a list of application types to be replaced
application_types_to_replace = application_type_counts[application_type_counts < 500].index.tolist()

# Replace in dataframe
for app in application_types_to_replace:
    application_df['APPLICATION_TYPE'] = application_df['APPLICATION_TYPE'].replace(app, "Other")

# Check to make sure replacement was successful
application_df['APPLICATION_TYPE'].value_counts()
```

```python
# Look at CLASSIFICATION value counts to identify and replace with "Other"
classification_counts = application_df['CLASSIFICATION'].value_counts()
print(classification_counts)

```

```python
# You may find it helpful to look at CLASSIFICATION value counts >1
# Look at CLASSIFICATION value counts greater than 1
classification_counts = application_df['CLASSIFICATION'].value_counts()
classification_counts[classification_counts > 1]
```

```python
# Look at CLASSIFICATION value counts
classification_counts = application_df['CLASSIFICATION'].value_counts()

# Choose a cutoff value: keep classifications with >100 occurrences
classifications_to_replace = classification_counts[classification_counts < 100].index.tolist()

# Replace in dataframe
for cls in classifications_to_replace:
    application_df['CLASSIFICATION'] = application_df['CLASSIFICATION'].replace(cls, "Other")

# Check to make sure replacement was successful
application_df['CLASSIFICATION'].value_counts()
```

```python
# Convert categorical data to numeric with `pd.get_dummies`
# Convert categorical data to numeric using one-hot encoding
application_df = pd.get_dummies(application_df)
```

```python
# Define target variable (y) and features (X)
y = application_df['IS_SUCCESSFUL']
X = application_df.drop('IS_SUCCESSFUL', axis=1)

# Split data into training and testing sets (e.g., 75% train, 25% test)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42, stratify=y
)
```

```python
# Create a StandardScaler instances
scaler = StandardScaler()

# Fit the StandardScaler
X_scaler = scaler.fit(X_train)

# Scale the data
X_train_scaled = X_scaler.transform(X_train)
X_test_scaled = X_scaler.transform(X_test)
```

## Compile, Train and Evaluate the Model

```python
# Number of input features
input_features = X_train.shape[1]

# Define the model
nn = tf.keras.models.Sequential()

# First hidden layer with 80 neurons and ReLU activation
nn.add(tf.keras.layers.Dense(units=80, activation='relu', input_dim=input_features))

# Second hidden layer with 30 neurons and ReLU activation
nn.add(tf.keras.layers.Dense(units=30, activation='relu'))

# Output layer with 1 neuron and sigmoid activation for binary classification
nn.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))

# Check the structure of the model
nn.summary()
```

```python
# Compile the model
nn.compile(
    loss='binary_crossentropy',   # Suitable loss function for binary classification
    optimizer='adam',             # Popular optimizer
    metrics=['accuracy']          # Track accuracy during training
)
```

```python
checkpoint_callback = tf.keras.callbacks.ModelCheckpoint(
    filepath='weights.{epoch:02d}.weights.h5',  # change extension here
    save_weights_only=True,
    save_freq='epoch',  # save every epoch (we'll handle saving every 5 epochs differently)
    verbose=1
)
class CustomCheckpoint(tf.keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        if (epoch + 1) % 5 == 0:  # Save every 5 epochs
            self.model.save_weights(f'weights.{epoch + 1:02d}.weights.h5')
            
checkpoint_callback = CustomCheckpoint()


```

```python
# Convert bool columns to int (0 or 1)
X_train = X_train.astype(int)

# Convert to NumPy array of float32 (preferred for TF)
X_train = X_train.values.astype('float32')

# Convert y_train to NumPy float32 array as well
y_train = y_train.values.astype('float32')

fit_model = nn.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=32,
    validation_split=0.2,
    callbacks=[checkpoint_callback],
    verbose=2
)

```

```python
# Evaluate the model using the test data
model_loss, model_accuracy = nn.evaluate(X_test_scaled,y_test,verbose=2)
print(f"Loss: {model_loss}, Accuracy: {model_accuracy}")
```
